- **What it shows:**  The Marital_Status column contains 749 rows where

```python
df.value_counts("Income_Category")
```

- **What I did:**  Executed .value_counts() against the DataFrame, using
Income_Category as parameter.
- **Why I did it:**  To show the count of each value within Income_Category to
determine if there were any unknowns.
- **What it shows:**  The Income_Category column contains 1112 rows where
Income_Category is 'Unknown'.

**How to treat 'Unknown' values?:**  Options include a. deleting rows that
contain an 'Unknown' value, b. replacing 'Unknown' with the most common value,
or c. keeping 'Unknown' as its own category.

A best case scenario where unknown values always occurred on the same row as
other unknown values would result in approximately 15.0% of rows with one or
more missing values (1519/10127 = 0.1500).

A worst case scenario where unknown values never occurred on the same row as
other unknown values would result in approximately 33.4% of rows with a missing
value ((1519+749+1112)/10127 = 0.3337).

Even for the best case scenario, losing just 15% of the data would lose
significant information from other 'complete' columns in any deleted data.

The second option, replacing 'Unknown' with the most common value would bias
the data towards one category which is unlikely to reflect the missing data.

Also, it's possible that a customer who doesn't provide a value for a
particular column may be less engaged with the bank, and may be more likely to
churn.  Keeping the data allows us to check that.

For these reasons I have chosen to keep 'Unknown' as its own category.  I
believe this maintains transparency about the data.  


```python
(df[["Education_Level", "Marital_Status", "Income_Category"]] == "Unknown").any(axis=1).sum()
```

- **What I did:**  Created a subset of columns from the DataFrame and tested
whether any column contained an 'Unknown' value, before summing up those
true unknowns.
- **Why I did it:**  To determine the true count of rows containing an
'Unknown' value.  Claude AI suggested that, rather than interpreting individual
column counts of 'Unknown' values to give a range (15.0% to 33.4%) for the true
number of rows containing one or more 'Unknown' values, I could instead
calculate the actual number of rows containing at least one 'Unknown' directly.
- **What it shows:**  3046 rows contain at least one column with an 'Unknown'
value.  This represents 30.1% of the total rows (3046/10127 = 0.3008).
Deleting that much data would likely impair any subsequent analysis.

- **Bug:**  My initial code for determining the “Unknown” row count didn't
work.  The traceback showed that an AttributeError was raised.
- **Cause:**  The code was incorrectly formatted and attempted to attach the
- .any() method to a string value, as well as attempting to create three
separate True/False series, instead of one DataFrame, for the three columns of interest.
- **Fix:**  Applied double square bracket syntax to code the columns and create
the DataFrame, as well as using parentheses to delimit the DataFrame code, so
the comparison runs before .any().
- **Lesson:**  Use double square brackets when working with multi-column lists,
and attach DataFrame methods directly to the DataFrame, using parentheses when
necessary.

### 2.4 Create the churn flag

```python
df['Attrition_Flag'].value_counts()
```

```python
df['Churn'] = df['Attrition_Flag'].map({'Attrited Customer': 1, 'Existing Customer': 0})
```

```python
df['Churn'].value_counts()
```

- **What I did:**  Created a new 'Churn' column that categorically encodes the
text strings in column 'Attrition_Flag'.
- **Why I did it:**  To provide data in a form suitable for analysis where the
data is required to be numeric because string data is unsuitable.
- **What it shows:**  The .value_counts() method applied to the 'Attrition_Flag'
column shows the count for each value within the column.  The same method
applied to the 'Churn' column shows that the value counts within this column
correspond to those in the 'Attrition_Flag' column.  This indicates that the
encoding was successful.
